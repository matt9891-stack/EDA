# Outliers Identification

```python
fig,axis =plt.subplots(1,2,figsize=(8,5))
sns.boxplot(df_red['fixed acidity'],orient = 'h',ax=axis[0])
sns.histplot(df_red['fixed acidity'],kde =True,ax=axis[1])
plt.tight_layout()
```

```python
Q1 = df_red['fixed acidity'].quantile(0.25)
Q3 = df_red['fixed acidity'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_red[(df_red['fixed acidity'] < lower_bound) | (df_red['fixed acidity'] > upper_bound)]
print("Outliers in 'fixed acidity':")
print(outliers)
```

# Remove Highly correlated features

```python
# Assume df is your dataframe and 'quality' is your target variable
X = df_red.drop(columns=["quality"])  # Predictors only

# Compute pairwise correlation matrix (absolute values)
corr_matrix = X.corr().abs()

# Select upper triangle of correlation matrix (to avoid double counting pairs)
upper_tri = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# Identify features with a correlation higher than a threshold (e.g., 0.80)
threshold = 0.80
to_drop_multicollinear = [
    column
    for column in upper_tri.columns
    if any(upper_tri[column] > threshold)
]

print("Features dropped due to high inter-correlation:", to_drop_multicollinear)
df_reduced = df_red.drop(columns=to_drop_multicollinear)
```

```python
# Compute correlation of all predictors with the target variable 'quality'
target_corr = X.corrwith(df_red["quality"]).abs()

# Set a minimum threshold for linear relationship strength (e.g., 0.05)
min_corr_threshold = 0.05
to_drop_low_target = target_corr[target_corr < min_corr_threshold].index.tolist()

print("Features dropped due to low correlation with target:", to_drop_low_target)

# Drop both sets of features
all_to_drop = list(set(to_drop_multicollinear + to_drop_low_target))
df_final = df_red.drop(columns=all_to_drop)
```
